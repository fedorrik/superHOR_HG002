```python
import numpy as np
from collections import Counter, defaultdict
from os import listdir


def get_list(file):
    out = []
    with open(file) as f:
        for line in f:
            line = line.strip().split()
            if len(line) > 0:
                out.append(line[3].split('::')[1][1:])
    return(np.array(out))


def min_cyclic_shift(s):
    """Находит лексикографически минимальный циклический сдвиг строки."""
    n = len(s)
    s = s + s  # Дублируем строку, чтобы искать циклический сдвиг
    return min(s[i:i+n] for i in range(n))


def sum_cyclic_permutations(d):
    grouped = defaultdict(int)
    
    for key, value in d.items():
        norm_key = min_cyclic_shift(key)  # Находим минимальный циклический сдвиг
        grouped[norm_key] += value        # Суммируем значения
    
    return dict(grouped)


def h_mer_percent(file, h):
    horhap_array = get_list(file)
    
    while len(horhap_array) % h != 0:
        horhap_array = horhap_array[:-1]
        
    h_mers = horhap_array.reshape(-1, h)
    h_mers_cnt = Counter([''.join(i) for i in h_mers])
    h_mers_cnt = Counter(sum_cyclic_permutations(h_mers_cnt)).most_common()
    h_mers_cnt = [('C' + '-C'.join(list(i[0])), i[1]) for i in h_mers_cnt]
    repeat_freq = h_mers_cnt[0][1] / len(h_mers)
    #print('best repeat freq: ', repeat_freq)
    for i in h_mers_cnt:
        print(i[0], i[1], sep='\t')
        
```

```python
for chrom in ['chr' + str(i) for i in range(1, 23)] + ['chrX', 'chrY']:
    
    file = 'stv_per_chr/{}_MATERNAL.bed'.format(chrom)
    print(file.split('/')[1].split('.')[0])
    try:
        h_mer_percent(file, 3)
    except FileNotFoundError:
        pass
    
    file = 'stv_per_chr/{}_PATERNAL.bed'.format(chrom)
    print(file.split('/')[1].split('.')[0])
    try:
        h_mer_percent(file, 3)
    except FileNotFoundError:
        pass
```
