```python
def get_superHOR(bed, n, repeat_number_threshold):

    # color for n-mer
    colors = ["182,14,16", "35,51,65", "209,226,240", "96,152,182", "152,113,67", "214,130,140", "201,173,153", "62,152,150"]
    color = colors[n-2]
    
    # trim array at the end
    while len(bed) % n != 0:
        bed = bed[:-1]
    
    chrom = bed[0][0]
    prev_nmer_bed = bed[:n]
    prev_nmer_str = ''.join([i[3] for i in prev_nmer_bed])
    i = 0
    status = 'no repeat'
    superhor_bed = []
    while i <= len(bed) - 2*n:
        i += n
        nmer_bed = bed[i:i+n]
        nmer_str = ''.join([i[3] for i in nmer_bed])
        if nmer_str != prev_nmer_str:
            # write prev repeat
            if status == 'repeat':
                if len(set([s for s in list(prev_nmer_str) if s.isdigit()])) > 1 and repeat_number >= repeat_number_threshold and prev_nmer_str[:n] != prev_nmer_str[n:]:
                    if n == 6 and prev_nmer_str[:4] == prev_nmer_str[4:8] == prev_nmer_str[8:]:
                        continue
                    if n == 9 and prev_nmer_str[:6] == prev_nmer_str[6:12] == prev_nmer_str[12:]:
                        continue
                    superhor_bed.append([prev_nmer_bed[1][0], repeat_start, repeat_end, f'{prev_nmer_str}::{repeat_number}', int(len(prev_nmer_str)/2), prev_nmer_bed[1][5], repeat_start, repeat_end, color])
                status = 'no repeat'
            repeat_number = 0
            repeat_start = nmer_bed[0][1]
            prev_nmer_bed = nmer_bed
            prev_nmer_str = nmer_str
        else:
            if status == 'no repeat':
                status = 'repeat'
                repeat_start = prev_nmer_bed[0][1]
                repeat_end = nmer_bed[-1][2]
                repeat_number = 2
            else: # status == 'repeat'
                repeat_end = nmer_bed[-1][2]
                repeat_number += 1
    
    return superhor_bed     
```

```python
# parse input horhap annotation
bed = []
with open('../HG002_horhap.v2.1_default-colors.bed') as f:
    for line in f:
        line = line.strip().split()
        line[3] = line[3].split('::')[1]
        bed.append(line)
```

```python
# run superhor function and write output superhor annotation
all_superhors = []
for n in range(2,10):
    superhor_bed = get_superHOR(bed, n, 3)
    all_superhors += superhor_bed
    with open(f'superHOR_{n}mer.bed', 'w') as f:
        f.write('track name="superHOR {}mer" description="superHOR {}mer" itemRgb="On" visibility="1"\n'.format(str(n), str(n)))
        for line in superhor_bed:
            f.write(f'{'\t'.join(list(map(str, line)))}\n')

with open(f'superHORs.bed', 'w') as f:
    for line in all_superhors:
        f.write(f'{'\t'.join(list(map(str, line)))}\n')
```

```python
# calculate proportion of array covered by superHOR
for chrom in set([i[0] for i in bed]):
    
    array_start = int([i[1] for i in bed if i[0] == chrom][0])
    array_end = int([i[2] for i in bed if i[0] == chrom][-1])
    array_length = array_end - array_start
    total_super_HOR_length = sum([int(i[2]) - int(i[1]) for i in all_superhors if i[0] == chrom])
    super_HOR_ratio = total_super_HOR_length / array_length
    print(chrom, super_HOR_ratio, sep='\t')
```
